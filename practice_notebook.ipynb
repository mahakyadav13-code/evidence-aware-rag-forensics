```python
import pandas as pd

df = pd.read_csv("chat_logs.csv")

# filter: messages from Alice
alice_msgs = df[df['sender'] == 'Alice']

# group: message count per sender
sender_counts = df.groupby('sender').size()

# sort by timestamp
df['timestamp'] = pd.to_datetime(df['timestamp'])
df_sorted = df.sort_values('timestamp')

print(alice_msgs)
print(sender_counts)
print(df_sorted)
```
